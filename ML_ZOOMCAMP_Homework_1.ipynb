```python
#Importing libraries to use
import pandas as pd
import numpy as np
```

```python
#Reading in the dataset
homework_data='data/car_fuel_efficiency.csv'
df1=pd.read_csv(homework_data)
df1
```

```python
#Question2
#Records count
number_of_rows=len(df1)
print(number_of_rows)
```

```python
#Question3
#Fuel types
fuel_types=df1['fuel_type'].nunique()
print(fuel_types)
```

```python
#Question4
#Number of columns with missing values
missing_values=df1.isnull().sum()
number_of_columns_with_missing_values=missing_values[missing_values>0]
print(len(number_of_columns_with_missing_values))
```

```python
#Question5
#Maximum fuel efficiency for cars from Asia
#
#getting a data frame for only Asian cars
asian_cars=df1[df1['origin']=='Asia']
asian_cars
#the maximum fuel efficiency for Asian cars
max_fuel_efficiency=asian_cars['fuel_efficiency_mpg'].max()
print(max_fuel_efficiency)
```

```python
#Question6
#Median of horsepower column
median=df1['horsepower'].median()
print('Median before filling in missing values is 149')
Median=149
#most frequent horsepower value
most_frequent=df1['horsepower'].mode()[0]
print(f'Most frequent value is {most_frequent}')
#filling in missing values with the most frequent value in that column
df1['horsepower']=df1['horsepower'].fillna(most_frequent)
print(df1)
new_median=df1['horsepower'].median()
New_median=new_median
print(f'New median is {New_median}')
if New_median > Median :
    print("Yes,it increased")
else:
    print("No,it decreased")
```

```python
#Question7
#Sum of weights
#1
#selecting only cars from Asia
asian_cars
#2
#working with the weight and model year columns
columns=asian_cars[['vehicle_weight','model_year']]
print(columns)
#3
#selecting the first 7 rows
values=columns.iloc[:7]
print(values)
```

```python
#Question7 cont'
#4
#convert to a Numpy array
X=np.array(values)
print(X)
```

```python
#5
#Transpose of X
XT=X.T
print(XT)
```

```python
#5
#Multiply X transpose and X
XTX=XT.dot(X)
print(XTX)
```

```python
#6
#Inverse of XTX
inverse=np.linalg.inv(XTX)
print(inverse)
```

```python
#7
#Create an array
Y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(Y)
```

```python
#8
#Multiplication
Z=inverse.dot(XT)
W=Z.dot(Y)
print(W)
```

```python
#9
#sum of the elements of W
sum=np.sum(W)
print(sum)
```
