```python
#working with plotly
#importing the right modules
from plotly.graph_objs import Scatter
from plotly import offline
import pandas as pd
import numpy as np 
import math
x=np.arange(0,math.pi*4,0.05)
y=np.sin(x)
data=[{'type':'scatter',
      'x':x,
      'y':y}]
offline.plot({'data':data},filename='Scatter.html')
```

```python
#using plotly.express for simple plots with less customization
import plotly.express as px
weekly_temps = pd.DataFrame({'day': ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
                             'temp': [28, 27, 25, 31, 32, 35, 36]})
print(weekly_temps)
fig = px.bar(data_frame=weekly_temps, x='day', y='temp')
#px is used by specifying a dataframe and using its columnsas arguments
fig.show()

```

```python
#bargraphs with plotly express
import pandas as pd
import plotly.express as px
data='data/mtn_customer_churn.csv'
df1=pd.read_csv(data)
df1.shape
print(df1.columns)
df2=df1.head()
df3=df1['Total Revenue'][0:15]
df3.to_excel('output.xlsx',index=False)
print(df3)
print(df2)
fig1=px.bar(data_frame=df2,x='Full Name',y='Number of Times Purchased')
fig1.show()
```

```python
#boxplot with plotly express
fig2=px.box(data_frame=df3,y='Total Revenue')
fig2.show()
```

```python
import pandas as pd
import plotly.express as px
weekly_temps=pd.DataFrame({'Day':['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
    'Temperature':[25,35,15,90,55,65,70],
    'Location':['Emi Pali','Emi Guni','Kpata','Shintaku','Shitel','Gboloko','Odenyi'] })
weekly_temps.head()
fig3=px.bar(data_frame=weekly_temps,
           x='Day',
           y='Temperature',
           color='Location',
           title='WEEKLY TEMPERATURES FOR BASSA LGA')
fig3.show()
```

```python
import pandas as pd
import plotly.express as px
peng='data/penguins.csv'
pengs=pd.read_csv(peng)
pengs=pengs.rename(columns={'Unnamed: 0':'Index'})
x1=pengs['island'][0:30]
y1=pengs['flipper_length_mm'][0:30]
z1=pengs['sex'][0:30]
#summary of the file
pengs.head()
```

```python
fig4=px.histogram(data_frame=pengs,x=x1,y=y1,color=z1,title='Penguins Analysis')
fig4.show()
```

```python
colour=dict(male='rgb(100,0,128)',female='black')
fig4=px.box(data_frame=pengs,y=y1,color=z1,color_discrete_map=colour,title='Boxplot')
fig4.show()
```
